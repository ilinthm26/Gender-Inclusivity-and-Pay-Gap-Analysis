# ***IBM Gender Inclusivity and Pay Gap Statistical Test***

# Import Libraries and Load Data

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df=pd.read_csv(r'/content/processed_hr_data_eda.csv')
```

# Prepare Data

```python
male_salary = df[df['Gender']=='Male']['MonthlyIncome']
female_salary=df[df['Gender']=='Female']['MonthlyIncome']

print("Male Sample Size:",len(male_salary))
print("Female Sample Size:",len(female_salary))
```

## T-Test (Welch's t-test)



```python
t_stat,p_value=stats.ttest_ind(
    male_salary,
    female_salary,
    equal_var=False #safer when variance differ
)

print("T-Statistics:",round(t_stat,4))
print("P-Value:",round(p_value,6))
```

Hypothesis Testing Result:

```python
alpha=0.05

if p_value < alpha:
    print("✅ p < 0.05 → Reject H0")
    print("Conclusion: The salary difference is statistically significant.")
else:
    print("❌ p ≥ 0.05 → Fail to Reject H0")
    print("Conclusion: No statistically significant salary difference detected.")
```

# Effects Size / Cohen's d
| Cohen’s d | Meaning      |
| --------- | ------------ |
| 0.2       | Small effect |
| 0.5       | Medium       |
| 0.8       | Large        |



```python
mean_diff=male_salary.mean()-female_salary.mean()
pooled_std=np.sqrt(
    ((male_salary.std()**2)+(female_salary.std()**2))/2
)
cohen_d=mean_diff/pooled_std

round(cohen_d,3)
```



An independent Welch’s t-test was conducted to evaluate whether salary differences exist between male and female employees.

The test produced a t-statistic of -1.22 and a p-value of 0.22.

Since the p-value exceeds the 0.05 threshold, we fail to reject the null hypothesis. This indicates that the observed salary difference is not statistically significant.

Furthermore, the effect size (Cohen’s d = -0.065) suggests the practical difference between genders is negligible.

Overall, the analysis does not provide evidence of a meaningful gender-based salary disparity within this dataset.
