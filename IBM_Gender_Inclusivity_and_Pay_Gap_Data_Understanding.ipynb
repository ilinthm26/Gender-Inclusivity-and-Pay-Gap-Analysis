# Feature Engineering

For HR D&I analytics, the most important engineered features are:

Age Bands

Tenure Bands

Leadership Flag

## Age Bands


```python
df['Age'].describe() # range from 18-60
```


Age Bands helps in:

Diversity analysis,

Attrition trend analysis

Leadership *Maturity*


```python
age_bins=[18,25,35,45,55,65]
age_labels=['18-25','26-35','36-45','46-55','56-65']
df['AgeBand']=pd.cut(
    df['Age'],
    bins=age_bins,
    labels=age_labels,
    right=False
)
```

Most of the employees,i.e., 37% are aged between 26-35. Followed by 34% between 36-45.

```python
df['AgeBand'].unique()
```

```python
df['AgeBand'].value_counts(normalize=True)*100
```

## Tenure Bands

```python
df['TotalWorkingYears'].describe() # range from 0 to 40
```

Tenure bands helps in analysing:

Are new hires leaving more?

Is leadership dominated by long-term employess?

Where is promotion stagation happening?

```python
tenure_bins=[0,2,5,10,20,40]
tenure_labels=['0-2 Years','3-5 Years','6-10 Years','11-20 Years','21-40 Years']

df['TenureBand']=pd.cut(
    df['YearsAtCompany'],
    bins=tenure_bins,
    labels=tenure_labels,
    right=True,
    include_lowest=True
)
```

31% of employees are with the company for last 10 years. We observe a decreasing percentage of employees in the longer tenure bands, with only 4.6% having 21-40 years of service. This could indicate a younger company, natural retirement rates, or potentially higher turnover in earlier career stages.

```python
df['TenureBand'].unique()
```

```python
df['TenureBand'].value_counts(normalize=True)*100
```

## Leadership Flags

```python
df['JobRole'].unique()
```

Leadership roles involved in the company are 'Manager',
    'Manufacturing Director',
    'Research Director' .
 22% of the employees are performing manager role in the organization.

```python
leadershipRoles=[
    'Manager',
    'Manufacturing Director',
    'Research Director'
 ]

df['IsLeader']=df['JobRole'].isin(leadershipRoles).astype(int)
df['IsLeader'].mean()
```

Feature engineering was performed to create structured workforce segments:

Age_Band enables demographic diversity analysis

Tenure_Band enables retention lifecycle analysis

Is_Leader simplifies leadership representation metrics
